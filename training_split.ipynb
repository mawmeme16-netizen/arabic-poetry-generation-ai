```python
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
final_df = pd.read_parquet(
    "../data/processed/final_poetry_dataset.parquet"
)
```

```python
print(final_df.shape)
```

```python
train_df, val_df = train_test_split(
    final_df,
    test_size=0.05,
    random_state=42,
    stratify=final_df["theme"]
)
```

```python
# نفحص
print(train_df.shape)
print(val_df.shape)
```

```python
# export train
train_df.to_json(
    "../data/processed/train.jsonl",
    orient="records",
    force_ascii=False,
    lines=True
)
```

```python
# export validation
val_df.to_json(
    "../data/processed/val.jsonl",
    orient="records",
    force_ascii=False,
    lines=True
)
```
