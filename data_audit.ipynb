### المكتبات وتحميل الملفات

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from tqdm import tqdm
import json
```

```python
DATA_DIR = Path("../data/raw")

JSON_PATH = DATA_DIR / "poems_dataset.json"
PARQUET_PATH = DATA_DIR / "train-00000-of-00002.parquet"
PARQUET_PATH1 = DATA_DIR / "train-00001-of-00002.parquet"
EXCEL_PATH = DATA_DIR / "Ashaar_Final_Clean.xlsx"
```

```python
# تحميل JSON
with open(JSON_PATH, "r", encoding="utf-8") as f:
    json_data = json.load(f)

json_df = pd.DataFrame(json_data)

print("JSON Shape:", json_df.shape)

json_df.head()
```

```python
# تحميل parquet الأول
parquet_df_1 = pd.read_parquet(PARQUET_PATH)

print("Parquet 1 Shape:", parquet_df_1.shape)



#تحميل parquet الثاني
parquet_df_2 = pd.read_parquet(PARQUET_PATH1)

print("Parquet 2 Shape:", parquet_df_2.shape)
```

```python
parquet_df_1.head()
```

```python
parquet_df_2.head()
```

```python
# دمجهم
parquet_df = pd.concat(
    [parquet_df_1, parquet_df_2],
    ignore_index=True
)

print("Merged Parquet Shape:", parquet_df.shape)
```

```python
# تحميل excel
excel_df = pd.read_excel(EXCEL_PATH)

print("Excel Shape:", excel_df.shape)

excel_df.head()
```
